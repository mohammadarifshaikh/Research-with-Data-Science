```python
file_path = '/content/drive/MyDrive/datasets/gym_membership.csv'
gym = pd.read_csv(file_path)  # No header in raw CSV
print("Dataset loaded successfully. Shape:", gym.shape)
```

 Bar chart of membership types — `gym['abonoment_type'].value_counts()`


```python
abonoment_counts = gym['abonoment_type'].value_counts()

fig, ax = plt.subplots()

ax.bar(abonoment_counts.index, abonoment_counts.values, color=['#F00', '#0000FF'])
ax.set(title='Membership Types', ylabel='Count')
plt.show()
```

Histogram of avg_time_in_gym


```python
fig, ax = plt.subplots()
ax.hist(gym['avg_time_in_gym'], bins = 20, color = 'seagreen', edgecolor = 'white')

ax.set(title='Average Time in Gym', ylabel='Count')
plt.show()
```

Scatter plot of Age vs visit_per_week

```python
fig, ax = plt.subplots()
ax.scatter(gym['Age'], gym['visit_per_week'], alpha = 0.4, s=18)

ax.set(title='Age vs. Visit Per Week', xlabel='Age', ylabel='Visit Per Week')
plt.show()
```

***Seaborn***

1. Box plot — `gender` vs `Age`

```python
fig, ax = plt.subplots(figsize=(11, 4))

sns.boxplot(data=gym, x='gender', y='Age', ax=ax)
ax.set_title('Age Distribution by Gender')
ax.set_xlabel('Gender')
ax.set_ylabel('Age')

plt.tight_layout()
plt.show()
```

2. Count plot — `days_per_week`


```python
fig, ax = plt.subplots( figsize=(8, 6))
sns.countplot(data=gym, x='days_per_week', ax=ax)
ax.set_title('Days Per Week')
ax.set_xlabel('Days Per Week')
ax.set_ylabel('Count')

ax.tick_params(axis='x', rotation=30)
plt.tight_layout()
plt.show()
```

3. Bar plot — mean `visit_per_week` per `abonoment_type`

```python

fig, ax = plt.subplots(figsize=(7, 4.5))

order = gym.groupby('abonoment_type')['visit_per_week'].mean().sort_values().index

sns.barplot(data=gym, x='abonoment_type', y='visit_per_week', order=order, ax=ax)

ax.set_title('Average Weekly Visits by Membership Type')
ax.set_xlabel('Membership Type (Abonoment Type)')
ax.set_ylabel('Mean Visits per Week')


ax.set_axisbelow(True)

plt.tight_layout()
plt.show()



```
